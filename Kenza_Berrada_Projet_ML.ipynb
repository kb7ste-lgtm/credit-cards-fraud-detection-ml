Data pre-processing

```python
df_processed = df.copy()
from sklearn import preprocessing
robust_scaler_amount = preprocessing.RobustScaler()
robust_scaler_time = preprocessing.RobustScaler()
df_processed['amount_scaled'] = robust_scaler_amount.fit_transform(df_processed['Amount'].values.reshape(-1, 1))
df_processed['time_scaled'] = robust_scaler_time.fit_transform(df_processed['Time'].values.reshape(-1, 1))
df_processed.drop(['Time', 'Amount'], axis=1, inplace = True)
df_processed.head()
```

```python
#definition des x et y
X = df_processed.drop('Class', axis = 1)
y = df_processed['Class']
```

Train-test split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train,y_test = train_test_split(X,y,test_size=0.2, random_state=1,stratify=y)
```

```python
#utilisation du SMOTE pour equilibré le dataset d'entrainement
from imblearn.over_sampling import SMOTE
oversample = SMOTE(sampling_strategy=1, random_state=1)
X_train_res, y_train_res = oversample.fit_resample(X_train, y_train)
print("results before SMOTE:\n", y_train.value_counts())
print("results after SMOTE:\n", y_train_res.value_counts())
```

Entrainement des models

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier
import xgboost as xgb
import lightgbm as lgb
```

```python
models = {
    'NB' : GaussianNB(),
    'LR' : LogisticRegression(max_iter=1000),
    'KNN': KNeighborsClassifier(),
    'SVM': LinearSVC(max_iter=1000, dual = False),
    'DT' : DecisionTreeClassifier(random_state = 1),
    'RF' : RandomForestClassifier(random_state=1, n_jobs=-1),
    'AB' : AdaBoostClassifier(random_state=1),
    'GB' : GradientBoostingClassifier(random_state=1),
    'XGB' : xgb.XGBClassifier(random_state=1, eval_metric='logloss'),
    'LGBM' : lgb.LGBMClassifier(random_state=1, verbose=-1)
}
```

```python
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, recall_score, precision_score, matthews_corrcoef, confusion_matrix, classification_report
```

```python
models_metrics={}
for model_name in models :
    model=models[model_name]
    model.fit(X_train_res,y_train_res)
    yhat=model.predict(X_test)
    models_metrics[model_name]=[round(accuracy_score(y_test,yhat),3), round(precision_score(y_test, yhat), 3), round(recall_score(y_test,yhat),3), round(f1_score(y_test,yhat),3),round(roc_auc_score(y_test,yhat),3), round(matthews_corrcoef(y_test, yhat),3) ]
print(models_metrics)
```

```python
dmetrics=pd.DataFrame(models_metrics,index=['Accuracy', 'Precision','Recall','F1-score', 'AUC', 'MCC'])
dmetrics.T
```

```python
def mc_report(model) :
    model.fit(X_train_res, y_train_res)
    yhat = model.predict(X_test)
    print(classification_report(y_test, yhat))
for model_name in models : 
    print('\n---------------------'+model_name+'------------------------')
    mc_report(models[model_name])
```

```python
#confusion matrix
def mc(model) : 
    yhat = model.predict(X_test)
    print('\n---------------Train------------------')
    print(confusion_matrix(y_train_res, model.predict(X_train_res)))
    print('\n-------------------------Test---------------------')
    print(confusion_matrix(y_test, yhat))

for model_name in models : 
    print('\n*********************'+model_name+'**************************')
    mc(models[model_name])
```

```python
from sklearn.metrics import RocCurveDisplay
plt.figure(figsize=(8,9))
ax = plt.gca()

for model_name, model in models.items():
    if model_name == 'SVM' : 
        continue
    RocCurveDisplay.from_estimator(model, X_test, y_test, ax=ax, name=model_name)
plt.title('Courbe ROC - Tous les modèles')
plt.show()
```

```python
from imblearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_val_score
cv_results=[]
cv=StratifiedKFold(n_splits=3, shuffle=True, random_state=1)

for model_name, model in models.items():
    pipeline = Pipeline(steps=[
        ('smote', SMOTE(random_state=1)),
        ('model', model)
    ])
    res = cross_val_score(pipeline, X_train, y_train,
    cv=cv, scoring='recall', n_jobs=-1
    )
    cv_results.append(res)
    print(f"{model_name} : {res.mean() :.3f} ± {res.std():.3f}")
    
```

```python
%matplotlib inline
plt.figure(figsize=(12,8))
plt.boxplot(cv_results)
plt.ylabel('Recall')
plt.title('Cross-validation Recall par modèle')
plt.xticks(ticks=np.arange(1,11), labels=models.keys(), rotation = 45, fontsize = 11)
plt.tight_layout()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

Tunning

```python
#librairies du tuning

from imblearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold, validation_curve
from sklearn.metrics import classification_report, confusion_matrix, f1_score
import pickle
```

```python
#paramètres des modèles - Random forest
print('paramètres du RF : ')
print(models['RF'].get_params())
```

```python
#paramètres des modèles - XGBoost
print('paramètres du XGB : ')
print(models['XGB'].get_params())
```

```python
#paramètres des modèles - Light GBM
print('paramètres du LGBM : ')
print(models['LGBM'].get_params())
```

```python
cvt = StratifiedKFold(n_splits=3, shuffle=True, random_state=1) #cross_validation du tuning
```

```python
RF_vc = Pipeline([('smote', SMOTE(random_state=1)), ('model', RandomForestClassifier(random_state=1, n_jobs=-1))])
XGB_vc=  Pipeline([('smote', SMOTE(random_state=1)), ('model', xgb.XGBClassifier(random_state=1, eval_metric='logloss', n_jobs=-1))])
LGBM_vc=Pipeline([('smote', SMOTE(random_state=1)), ('model', lgb.LGBMClassifier(random_state=1, verbose=-1))])
```

```python
#validation curves: Random Forest
#1) RF
max_depth_liste = [i+2 for i in range(15)]
train_score, test_score = validation_curve(RF_vc, X_train, y_train, param_name='model__max_depth', param_range=max_depth_liste,cv=3, scoring='f1', n_jobs=-1)

plt.figure(figsize=(8,5))
plt.title("validation curve Random Forest : max_depth")
plt.xlabel("max_depth")
plt.ylabel("F1_Score")
plt.plot(max_depth_liste, train_score.mean(axis=1), label='Training', marker='o')
plt.plot(max_depth_liste, test_score.mean(axis=1), label='Test', marker='o')
plt.xticks(max_depth_liste)
plt.legend(loc="best")
plt.show()
```

```python
#validation curves: XGBOOST
learning_rate_liste = [0.01, 0.05, 0.1, 0.2, 0.3, 0.5]
train_score, test_score = validation_curve(XGB_vc, X_train, y_train, param_name='model__learning_rate', param_range=learning_rate_liste,cv=3, scoring='f1', n_jobs=-1)

plt.figure(figsize=(8,5))
plt.title("validation curve XGBOOST : learning rate")
plt.xlabel("learning_rate")
plt.ylabel("F1_Score")
plt.plot(learning_rate_liste, train_score.mean(axis=1), label='Training', marker='o')
plt.plot(learning_rate_liste, test_score.mean(axis=1), label='Test', marker='o')
plt.xticks(learning_rate_liste)
plt.legend(loc="best")
plt.show()
```

```python
#validation curves: Light GBM

train_score, test_score = validation_curve(LGBM_vc, X_train, y_train, param_name='model__learning_rate', param_range=learning_rate_liste,cv=3, scoring='f1', n_jobs=-1)

plt.figure(figsize=(8,5))
plt.title("validation curve Light GBM : learning rate")
plt.xlabel("learning_rate")
plt.ylabel("F1_Score")
plt.plot(learning_rate_liste, train_score.mean(axis=1), label='Training', marker='o')
plt.plot(learning_rate_liste, test_score.mean(axis=1), label='Test', marker='o')
plt.xticks(learning_rate_liste)
plt.legend(loc="best")
plt.show()
```

```python
#tuning Random Forest : 
#Pour le tunning des trois modèle j'ai choisi RandomizedSearch pour des raison d'éfficacité computationelle et avoir des reusltat et j'ai choisi le nombre d'iteration et mis n_jobs=-1 pour la meme raison
tuning_RF = Pipeline([('smote', SMOTE(random_state=1)),('model', RandomForestClassifier(random_state=1, n_jobs=-1))])
Param_grid= {'model__n_estimators' : [100], 'model__max_depth': [8,10,12], 'model__min_samples_split':[2,5,10]}
RF_search = RandomizedSearchCV(tuning_RF,Param_grid, n_iter=5, cv=cvt, scoring='f1', n_jobs=-1, random_state=1, verbose=2)
RF_search.fit(X_train, y_train)
print('Meilleurs paramètres RF:', RF_search.best_params_)
```

```python
#tuning XGBOSST
tuning_XGB= Pipeline([('smote', SMOTE(random_state=1)),('model', xgb.XGBClassifier(random_state=1, eval_metric='logloss', n_jobs=-1))])
Param_grid= {'model__n_estimators' : [100], 'model__max_depth': [3,5,7], 'model__learning_rate' : [0.01, 0.1,0.2], 'model__subsample' : [0.8, 1.0]}

XGB_search = RandomizedSearchCV(tuning_XGB, Param_grid, n_iter =5, cv=cvt, scoring='f1', n_jobs=-1, random_state=1, verbose=2)
XGB_search.fit(X_train, y_train)
print('Meilleurs paramètres XGB :', XGB_search.best_params_)
```

```python
#tuning LGBM
tuning_LGBM = Pipeline([('smote', SMOTE(random_state=1)),('model', lgb.LGBMClassifier(random_state=1, verbose=-1))])
Param_grid= {'model__n_estimators' : [100], 'model__max_depth': [3,5,7,-1], 'model__learning_rate' : [0.01, 0.1,0.2], 'model__num_leaves' : [31,50,100]}

LGBM_search = RandomizedSearchCV(tuning_LGBM, Param_grid, n_iter =5, cv=cvt, scoring='f1', n_jobs=-1, random_state=1, verbose=2)
LGBM_search.fit(X_train, y_train)
print('Meilleurs paramètres LGBM:', LGBM_search.best_params_)
```

```python
#evaluation des modèles sur le datset test
meilleur_RF = RF_search.best_estimator_
meilleur_XGB = XGB_search.best_estimator_
meilleur_LGBM = LGBM_search.best_estimator_

for name, model in {'RF' : meilleur_RF, 'XGB' : meilleur_XGB, 'LGBM' : meilleur_LGBM}.items():
    yhat=model.predict(X_test)
    print(classification_report(y_test, yhat))
    print(confusion_matrix(y_test, yhat))
```

```python
#comparaison de la performance avant vs après le tuning

tuning_résultats = {
    'RF' : [round(f1_score(y_test, models['RF'].predict(X_test)), 3),
            round(f1_score(y_test, meilleur_RF.predict(X_test)), 3)],
    'XGB' :[round(f1_score(y_test, models['XGB'].predict(X_test)), 3),
            round(f1_score(y_test, meilleur_XGB.predict(X_test)), 3)],
    'LGBM' :  [round(f1_score(y_test, models['LGBM'].predict(X_test)), 3),
            round(f1_score(y_test, meilleur_LGBM.predict(X_test)), 3)],
}

Comparatif = pd.DataFrame(tuning_résultats, index=['F1 avant tuning', 'F1 après tuning'])
Comparatif.T
   #Le light GBM est le meilleur model a prendre en considération         
```
